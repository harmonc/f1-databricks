```sql
create or refresh materialized view f1.silver.drivers_names
as
select 
  driverId,
  concat(forename,' ',surname) as name
from f1.bronze.drivers
```

```sql
CREATE OR REFRESH MATERIALIZED VIEW f1.silver.lap_times
SELECT 
  laps.raceId,
  laps.driverId,
  races.circuitId,
  laps.lap,
  laps.position,
  laps.time,
  laps.milliseconds,
  CAST(ROUND(laps.milliseconds/1000) as INT) as seconds,
  CAST(laps.milliseconds/1000 as INT) as seconds_int,
  FLOOR(CAST(laps.milliseconds/1000 as INT)/60) as minute_part,
  CAST(laps.milliseconds/1000 as INT)%60 as seconds_part,
  laps.milliseconds%1000 as ms_part
FROM f1.bronze.lap_times laps
JOIN f1.bronze.races races
ON races.raceId = laps.raceId
```

```sql
create or refresh materialized view f1.silver.circuit_names
as
select 
  circuitId as id,
  name
from f1.bronze.circuits
```

```sql
create or refresh materialized view f1.silver.drivers_names
as
select 
  driverId,
  concat(forename,' ',surname) as name
from f1.bronze.drivers
```

```sql
create or refresh materialized view f1.silver.lap_times
SELECT 
  raceId,
  driverId,
  lap,
  position,
  time,
  milliseconds,
  CAST(ROUND(milliseconds/1000) as INT) as seconds,
  CAST(milliseconds/1000 as INT) as seconds_int,
  FLOOR(CAST(milliseconds/1000 as INT)/60) as minute_part,
  CAST(milliseconds/1000 as INT)%60 as seconds_part,
  milliseconds%1000 as ms_part
FROM f1.bronze.lap_times laps
```

```sql
create or refresh materialized view f1.silver.circuit_names
as
select 
  circuitId as id,
  name
from f1.bronze.circuits
```

```sql
create or refresh materialized view f1.gold.race_results
as
select
  races.full_name,
  drivers.name,
  results.positionOrder as position,
  coalesce(nullif(regexp_replace(results.time, '\\\\N', ''), ''),status.status) as result
from f1.bronze.results results
join f1.silver.dim_races races on results.raceId = races.raceId
join f1.silver.drivers_names drivers on results.driverId = drivers.driverId
join f1.bronze.status status on results.statusId = status.statusId
```

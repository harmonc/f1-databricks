```python
df = spark.table("f1.bronze.lap_times")
```

```python
from pyspark.sql.functions import col

df = df.where(col("raceId") == 1)
display(df)
```

```python
import os

current_location = os.getcwd()
current_location
```

```python
import plotly.express as px

pdf = df.select("driverId","lap", "position").toPandas()
fig = px.line(pdf, x="lap", y="position",color="driverId", title="Lap Times (ms) by Lap")
display(fig)
```

```python
html_string = fig.to_html()

with open(f"{current_location}/my_html_file.html", "w") as f:
    f.write(html_string)
```

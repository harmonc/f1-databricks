```sql
SELECT 
time,
milliseconds,
CAST(ROUND(milliseconds/1000) as INT) as seconds,
CAST(milliseconds/1000 as INT) as seconds_int,
FLOOR(CAST(milliseconds/1000 as INT)/60) as minute_part,
CAST(milliseconds/1000 as INT)%60 as seconds_part,
milliseconds%1000 as ms_part
FROM f1.bronze.lap_times laps
ORDER BY milliseconds asc
```
